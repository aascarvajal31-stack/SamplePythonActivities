```python
import math
x1=float(input("Enter x1: "))
y1=float(input("Enter y1: "))
x2=float(input("Enter x2: "))
y2=float(input("Enter y2: "))
distance=math.sqrt((pow(x2-x1, 2))+(pow(y2-y1, 2)))
print("The distance between the two points is: ", distance)

# ***Reflection & Evaluation***

#A math library is more practical than writing all calculations from scratch because it saves you more time and is more simple. Writing this code would be much more difficult because I'd have to type the math formula manually insted of it being directly integrated and accessible with a simple function
```
