```python
# import the built in math library to access its predefined functions in order to simplify adding the formula later
import math

# Get the coordinates of the first point from the user
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

# Get the coordinates of the second point from the user
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

distance = math.sqrt((pow(x2-x1, 2))+(pow(y2-y1, 2)))
#import the

print("The distance between the two points is: ", distance)
#This displays the output of the distance formula, which is the final distance between the two coordinates
```
